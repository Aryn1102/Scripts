```python
df = pd.DataFrame(
    {
        "Name": ["Raj", "Aman", "Priya"],
        "Age": [22, 21, 23],
        "CGPA": [6.22, 7.8, 8.9],
        "weight": [55, 60, 58],
        "college": ["ABC", "XYZ", "PQR"],
        "skills": ["Python", "Java", "C++"]
    }
)
```
