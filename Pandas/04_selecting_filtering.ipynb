```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Raj", "Aman", "Priya", "Ravi"],
    "age": [22, 21, 23, 20],
    "cgpa": [6.2, 8.1, 7.5, 9.0]
})
print(df[df["cgpa"] > 7]["name"])
#aman, priya, ravi
```

```python
print(df.loc[1:2, ["name", "cgpa"]])
#rows: index 1, 2
#column: name, cgpa
```

```python
print(df.iloc[0:2, 0:2])
#rows: index 0, 1
#columns: name, age
```

```python
""" 
Find students with CGPA greater than 7.
Return only:
name
cgpa
"""
df.loc[df["cgpa"] > 7, ["name", "cgpa"]]
```

```python
""" 
Create a column:
passed
with:
CGPA >= 5 → True
CGPA < 5  → False
"""
df["passed"] = df["cgpa"] >= 5
```

```python
""" 
Create:
performance
using:
CGPA >= 8 → "Excellent"
CGPA >= 7 → "Good"
otherwise → "Average"
Use loc.
"""
df.loc[df["cgpa"] >= 8, "performance"] = "Excellent"
df.loc[(df["cgpa"] >= 7) & (df["cgpa"] < 8), "performance"]  = "Good"
df.loc[df["cgpa"] < 7, "performance"] = "Average"
df["performance"]
```

```python
""" 
Given:
Create:
salary_per_year
using:
salary / (experience + 1)
"""
df = pd.DataFrame({
    "age": [20, 21, 22, 23, 24],
    "salary": [25000, 30000, 35000, 40000, 45000],
    "experience": [0, 1, 2, 3, 4]
})
df["salary_per_year"] = df["salary"] / (df["experience"] + 1)
```
