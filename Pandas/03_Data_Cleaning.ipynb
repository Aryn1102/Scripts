```python
import pandas as pd
import numpy as np

df = pd.DataFrame(
    {
        "Name": ["Raj", "Aman", "Priya", "Raj", None],
        "Age": [22, 21, np.nan, 22, 20],
        "CGPA": [6.22, 7.8, 8.9, 6.22, np.nan],
        "Branch": ["IT", "CSE", "IT", "IT", None]
    }
)

print(df)
```

```python
print(df.isnull())
""" 
What does True mean?
it means that place is empty
"""
```

```python
print(df.isnull().sum())
""" 
which columns has the most missing values?
all are same
"""
```

```python
clean_df = df.dropna()

print(clean_df)
""" 
What happened to the rows containing missing values?
the rows have been dropped
"""
```

```python
df["Age"] = df["Age"].fillna(0)
df["Branch"] = df["Branch"].fillna("Unknown")
""" 
Why might filling with 0 be a bad idea for age in a real project?
it will affect the meadian and mean values of the age
"""
```

```python
print(df.duplicated())
""" 
which row is duplicated?
4"""
```

```python
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print(df)
""" 
how many rows remain?
4
"""
```

```python
data = {
    "Marks": [90, 80, None, 70, 90],
    "Branch": ["IT", "CSE", "IT", None, "IT"]
}
data_df = pd.DataFrame(data)
```

```python
print(data_df.isnull())
```

```python
print(data_df.isnull().sum())
```

```python
avg = data_df["Marks"].mean()
data_df["Marks"] = data_df["Marks"].fillna(avg)
data_df["Branch"] = data_df["Branch"].fillna("Unknown")
```

```python
data_df.duplicated().sum()
```

```python
data_df.drop_duplicates()
```
