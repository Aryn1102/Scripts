```python
import sqlite3
```

```python
conn = sqlite3.connect('academy.db')
cursor = conn.cursor()
"""
A cursor is created from a database connection.
It is used to execute SQL queries and retrieve results.
"""
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    cgpa REAL,
    branch TEXT
    )
    """)
conn.commit()
```

```python
cursor.execute("DELETE FROM students")
conn.commit()
cursor.execute("""
INSERT INTO students (id, name, age, cgpa, branch)
VALUES (1, 'raj', 22, 6.22, 'IT'),
       (2, 'Aman', 21, 7.8, 'CSE'),
       (3, 'Priya', 23, 8.9, 'IT'),
       (4, 'Ravi', 22, 7.1, 'CSE');
""")
conn.commit()
```

```python
cursor.execute("SELECT * FROM students")
print(cursor.fetchall())
```

```python
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("""
SELECT name, cgpa FROM students;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
