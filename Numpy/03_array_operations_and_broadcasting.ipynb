```python
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

print(a + b) #[5, 7, 9]
print(a - b) #[-3, -3, -3]
print(a * b) #[4, 10, 18]
print(a / b) #[0.25, 0.4, 0.5]
```

```python
arr = np.array([10,20,30])

print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 10)
""" 
Why doesn't NumPy need a loop here?
numpy doesnt need a loop here because every elements in the numpy array will be treated the same and braodcasted the same operations. 
"""
```

```python
numbers = [1,2,3,4]

result = []

for n in numbers:
    result.append(n*2)

print(result)
```

```python
""" 
What is vectorization?
vectorization is running operation on whole block of data at once rather than each element individually.

Why is NumPy faster than Python loops?
numpy is built upon c and fortran which makes it highly efficient.It is also faster because arrays store elements in contiguous memory,
 making operations more cache-friendly.

What is broadcasting?
NumPy automatically expands compatible arrays so arithmetic can be performed without manually copying data.

Difference between scalar operations and array operations?
scalar operation is applying operations on a single data element and array operations is applying the operation on all the elements of array at once

Why are NumPy mathematical functions preferred over writing loops?
it is faster due to vectorization.
"""
```

```python
import numpy as np

arr = np.array([2,4,6])

print(arr + arr) #[4, 8, 12]

print(arr * arr) #[4,16,36]

print(arr > 3) #[False, True, True]

print(arr[arr > 3]) #[4, 6]
```
