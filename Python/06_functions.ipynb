```python
""" 
Write:
def find_largest(numbers):
that returns the largest number without using max().
"""
def find_largest(numbers):
    a=numbers[0]
    for i in numbers:
        if i > a:
            a=i
    return a
```

```python
""" 
Write:
def count_even(numbers):
that returns the number of even elements.
"""
def count_even(numbers):
    count = 0
    for i in numbers:
        if i % 2 == 0:
            count+=1
    return count
```

```python
""" 
Write:
def reverse_list(numbers):
that returns the reversed list.
Don't use:
.reverse()
or:
[::-1]
"""
def reverse_list(numbers):
    reverse = []
    for i in range(len(numbers)):
        reverse.append(numbers.pop())
    return reverse
```

```python
""" 
Write:
def calculate_average(scores):
that returns the average of a list of scores without using sum().
Example:
calculate_average([80, 90, 70])
should return:
80
"""
def calculate_average(scores):
    a = 0
    count = 0
    for i in scores:
        a += i
        count +=1
    return (a/count)
```

```python
""" 
Write:
def second_largest(numbers):
that returns the second-largest distinct number without using sort() or sorted().
Example:
second_largest([10, 5, 20, 8, 20, 15])
Expected:
15
"""
def second_largest(numbers):
    largest = float('-inf')
    second = float('-inf')
    for i in numbers:
        if i > largest:
            second = largest
            largest = i
        elif i < largest and i >second:
            second = i
    return second
```
