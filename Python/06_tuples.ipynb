```python
numbers = (10, 20, 30)
def get_min_max(numbers):
    minimum = numbers[0]
    maximum = numbers[0]

    for number in numbers:
        if number < minimum:
            minimum = number
        if number > maximum:
            maximum = number

    return minimum, maximum

minimum, maximum = get_min_max(numbers)
```
