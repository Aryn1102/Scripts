```python
""" 
numbers = [1, 2, 2, 3, 1, 2, 4, 3, 3]
Create a dictionary containing the frequency of every number:
{
    1: 2,
    2: 3,
    3: 3,
    4: 1
}
Don't use collections.Counter.
This is particularly useful for DSA, so take this one seriously.
"""
numbers = [1, 2, 2, 3, 1, 2, 4, 3, 3]

count = {}

for number in numbers:
    if number in count:
        count[number] += 1
    else:
        count[number] = 1

print(count)
```
