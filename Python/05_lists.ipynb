```python
""" 
DSA — Two Sum
This is your first major LeetCode pattern from the list/array family.
Given:
numbers = [2, 7, 11, 15]
target = 9
Find two numbers whose sum is target and return their indices.
Expected:
[0, 1]
"""
numbers = [2, 7, 11, 15]
target = 9
indices  = []
for i in range(len(numbers)):
    for j in range(i+1, len(numbers)):
        if numbers[i] + numbers[j] == target:
            indices =[i, j]
            break
print(indices)
```
