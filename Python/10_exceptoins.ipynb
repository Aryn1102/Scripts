```python
""" 
Write:
def convert_age(age):
The function should:
convert age to an integer
return the integer
handle invalid input using ValueError
"""
def convert_age(age):
    try:
        return int(age)
    except ValueError:
        return "age is invalid"
```

```python
""" 
Write:
def get_score(scores, student):
Return the student's score.
If the student doesn't exist, handle the KeyError and return:
"Student not found"
"""
def get_score(scores, student):
    try:
        return scores[student]
    except KeyError:
        print("student doesnt exist")
```

```python
""" 
Write:
def calculate_average(numbers):
It should:
raise ValueError if numbers is empty
otherwise return the average
"""
def calculate_average(numbers):
    if not numbers:
        raise ValueError("numbers cannot be empty")

    return sum(numbers) / len(numbers)
```
