```python
""" 
Given:
python_students = {"Raj", "Aman", "Priya", "Ravi"}
ml_students = {"Raj", "Priya", "Karan", "Neha"}
Find:
Students taking either Python or ML
Students taking both
Students taking Python but not ML
Students taking ML but not Python
"""
python_students = {"Raj", "Aman", "Priya", "Ravi"}
ml_students = {"Raj", "Priya", "Karan", "Neha"}

print(python_students | ml_students)
print(python_students & ml_students)
print(python_students - ml_students)
print(ml_students - python_students)
```

```python
""" 
Write:
def has_duplicate(numbers):
It should return:
True
if the list contains a duplicate and:
False
otherwise.
Do not use set(numbers) directly for the solution.
Use a seen set and iterate through the list.
"""
def has_duplicate(numbers):
    seen = set()
    for number in numbers:
        if number in seen:
            return True
        seen.add(number)
    return False
```
