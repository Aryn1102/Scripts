```python
""" 
Given:
numbers = [1, 2, 3, 4, 5]
Create:
[1, 4, 9, 16, 25]
using a list comprehension.
"""
numbers = [1, 2, 3, 4, 5]
result = [num ** 2 for num in numbers]
```

```python
""" 
Given:
numbers = [10, 15, 20, 25, 30, 35]
Create a list containing only numbers divisible by 5 and greater than 20.
Expected:
[25, 30, 35]
"""
numbers = [10, 15, 20, 25, 30, 35]
result = [num for num in numbers if num%5==0 and num >20]
print(result)
```

```python
""" 
Given:
scores = [45, 60, 72, 38, 90]
Create:
["Fail", "Pass", "Pass", "Fail", "Pass"]
using a comprehension.
Passing means >= 50.
"""
scores = [45, 60, 72, 38, 90]
result = ["Pass" if num >= 50 else "Fail" for num in scores]
print(result)
```

```python
""" 
Given:
numbers = [1, 2, 3, 4, 5]
Create:
{
    1: 1,
    2: 4,
    3: 9,
    4: 16,
    5: 25
}
using a dictionary comprehension.
"""
numbers = [1, 2, 3, 4, 5]
result = { num: num ** 2 for num in numbers}
print(result)
```
