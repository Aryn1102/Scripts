```python
""" 
Create:
class Student:
It should accept name, age, and cgpa, store them as attributes, and print them.
Add:
def is_passed(self):
Return True if CGPA ≥ 5, otherwise False.
"""
class Student:
    def __init__(self, name, age, cgpa):
        self.name = name
        self.age = age
        self.cgpa = cgpa

    def is_passed(self):
        if self.cgpa >=5:
            return True
        else:
            return False
        
student1 = Student("Raj", 22, 5.88)
print(student1.name)
print(student1.age)
print(student1.cgpa)
```

```python
""" 
Create:
class DatasetLoader:
with load(), then create CSVLoader(DatasetLoader) and override load() to return "Loading CSV".
Create CSVLoader and JSONLoader, both with load() returning different messages.
Then write:
def process_data(loader):
that calls loader.load().
Use it with both classes.
"""
class DatasetLoader:
    def load(self):
        pass

class CSVLoader(DatasetLoader):
    def load(self):
        return "Loading CSV"
class JSONLoader(DatasetLoader):
    def load(self):
        return "Loading JSON"
def process_data(loader):
    return loader.load()

csv = CSVLoader()
json = JSONLoader()

print(process_data(csv))
print(process_data(json))
```

```python
""" 
Given:
student = {
    "name": "Raj",
    "age": 22,
    "cgpa": 6.2
}
Write code that:
saves it to student.json
reads it back
prints the student's name
"""
import json
student = {
    "name": "Raj",
    "age": 22,
    "cgpa": 6.2
}
with open("students.json", "w") as file:
    json.dump(student, file, indent=4)
with open("students.json", "r") as file:
    data = json.load(file)
    print(data["name"])
```
