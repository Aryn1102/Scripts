```python
salary = 70000

if salary > 50000:
    print("High Salary")

if salary < 50000:
    print("High Salary")

if salary == 50000:
    print("High Salary")

if salary != 50000:
    print("High Salary")

if salary <= 50000:
    print("High Salary")

if salary >= 50000:
    print("High Salary")
```
